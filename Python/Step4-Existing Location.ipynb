```python
import folium
import pandas as pd

# Load dataset
df = pd.read_csv("C:\\Users\\Yash\\OneDrive\\Desktop\\Yash\\Evision\\Evision\\Files\\Code Generated CSV\\Complete_Master_Data.csv")

# Create a map centered on the mean location
center_lat = df['latitude'].mean()
center_lon = df['longitude'].mean()
m = folium.Map(location=[center_lat, center_lon], zoom_start=10)

# Add detailed markers
for _, row in df.iterrows():
    popup_html = f"""
    <h4>{row['charging_station_name']}</h4>
    <b>Operator:</b> {row['charging_station_operator']}<br>
    <b>Location:</b> {row['charging_station_location']}<br>
    <b>City:</b> {row['city']}<br>
    <b>State:</b> {row['state']}<br>
    <b>Charger Type:</b> {row['Charger Type']}<br>
    <b>Battery Capacity:</b> {row['battery_capacity']} kWh<br>
    <b>Energy Consumed:</b> {row['energy_consumed']} kWh<br>
    <b>Charging Duration:</b> {row['charging_duration']} mins<br>
    <b>Charging Rate:</b> {row['charging_rate']} kW<br>
    <b>Busiest Hours:</b> {row['time_of_day']}<br>
    <b>Busiest Day of Week:</b> {row['day_of_week']}<br>
    <b>Payment Type:</b> {row['payment_type']}<br>
    <b>User Type:</b> {row['user_type']}
    """
    folium.Marker(
        location=[row['latitude'], row['longitude']],
        popup=folium.Popup(popup_html, max_width=350),
        tooltip=row['charging_station_name'],
        icon=folium.Icon(color="green", icon="bolt", prefix="fa")
    ).add_to(m)


# Save map to HTML file
m.save("Existing_EV_Charging_Stations"".html")
print("✅ Map saved as 'Existing_EV_Charging_Stations.html' — open it in a browser to view.")
```
