```python
import pandas as pd
import matplotlib.pyplot as plt

# Load datasets
df = pd.read_csv(r"C:\\Users\\Yash\\OneDrive\\Desktop\\Yash\\Evision\\Evision\\Files\\Cleaned dataset\\Complete_Master_Data.csv")
# Set style for plots
plt.style.use("ggplot")
```

```python
if "Charger Type" in df.columns:
    charger_counts = df["Charger Type"].value_counts()
else:
    charger_counts = pd.Series(dtype=int)

if not charger_counts.empty:
    plt.figure(figsize=(6,6))
    plt.pie(charger_counts, labels=charger_counts.index, autopct="%1.1f%%", startangle=140)
    plt.title("Charger Type Distribution")
    plt.show()
```

```python
# Energy Consumed
if "energy_consumed" in df.columns:
    plt.figure(figsize=(6,4))
    df["energy_consumed"].dropna().hist(bins=30, color="steelblue")
    plt.title("Distribution of Energy Consumed (kWh)")
    plt.xlabel("kWh")
    plt.ylabel("Frequency")
    plt.show()

# Charging Duration
if "charging_duration" in df.columns:
    plt.figure(figsize=(6,4))
    df["charging_duration"].dropna().hist(bins=30, color="orange")
    plt.title("Distribution of Charging Duration (Hours)")
    plt.xlabel("Hours")
    plt.ylabel("Frequency")
    plt.show()
```

```python
# Time of Day
if "time_of_day" in df.columns:
    plt.figure(figsize=(6,4))
    df["time_of_day"].value_counts().plot(kind="bar", color="purple")
    plt.title("Sessions by Time of Day")
    plt.xlabel("Time of Day")
    plt.ylabel("Count")
    plt.show()

# Day of Week
if "day_of_week" in df.columns:
    order = ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
    df["day_of_week"] = pd.Categorical(df["day_of_week"], categories=order, ordered=True)
    plt.figure(figsize=(8,4))
    df["day_of_week"].value_counts().reindex(order).plot(kind="bar", color="skyblue")
    plt.title("Sessions by Day of Week")
    plt.xlabel("Day")
    plt.ylabel("Count")
    plt.show()
```

```python
if "latitude" in df.columns and "longitude" in df.columns:
    plt.figure(figsize=(6,6))
    plt.scatter(df["longitude"], df["latitude"], alpha=0.6, c="blue", edgecolor="k")
    plt.title("EV Station Locations")
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.show()
```
